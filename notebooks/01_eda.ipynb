```python
###create new table for customers and finding now first and last purchase
orders["order_purchase_timestamp"] = pd.to_datetime(
    orders["order_purchase_timestamp"],
    format="mixed"
)
orders_customer = orders.merge(
    customers[['customer_id', 'customer_unique_id']],
    on='customer_id',
    how='left'
)

df1 = (
    orders_customer
    .groupby('customer_unique_id')
    .agg(
        total_orders=('order_id', 'nunique'),
        first_purchase=('order_purchase_timestamp', 'min'),
        last_purchase=('order_purchase_timestamp', 'max')
    )
    .reset_index()
)

df1.head()
```

```python
df1.shape
```

```python
###repeat customer findings
repeat_customers = df1[df1['total_orders'] > 1]

repeat_customers.shape
repeat_percentage = len(repeat_customers) / len(df1) * 100

repeat_percentage
```

```python
###Revenue analysis

payment_type_summary = (
    payments.groupby('payment_type')['payment_value']
    .agg(
        total='sum',
        average='mean',
        count='count'
    )
    .sort_values('total', ascending=False)
    .reset_index()
)

payment_type_summary.to_csv(
    r"C:\Users\Asus\Desktop\customer_prediction\data\processed\payment_type_revenue.csv",
    index=False
)
```

```python
##order status checking and calculating
order_status_summary = (
    orders.groupby('order_status')['order_id']
    .count()
    .reset_index(name='order_count')
)

order_status_summary['percentage'] = (
    order_status_summary['order_count']
    / order_status_summary['order_count'].sum()
    * 100
)

order_status_summary = order_status_summary.sort_values(
    'order_count',
    ascending=False
)

order_status_summary.to_csv(
    r"C:\Users\Asus\Desktop\customer_prediction\data\processed\order_status_summary.csv",
    index=False
)
```

```python
####review analysis
review_analysis = (
    reviews
    .groupby('review_score')
    .agg(
        review_count=('review_id', 'count'),
        avg_score=('review_score', 'mean')
    )
    .sort_index()
)

review_analysis
reviews['review_score'].value_counts().sort_index()
review_analysis.to_csv(
    'C:/Users/Asus/Desktop/customer_prediction/data/processed/review_analysis.csv',
    index=False
)
```

```python
orders["year"] = orders["order_purchase_timestamp"].dt.year
orders["month"] = orders["order_purchase_timestamp"].dt.month
```

```python
###yearly sales
yearly_sales = (
    orders.groupby('year')['order_id']
    .nunique()
)

yearly_sales
```

```python
###monthly sales
monthly_sales = (
    orders.groupby(['year', 'month'])['order_id']
    .nunique()
    .reset_index(name='total_orders')
)

monthly_sales
```

```python
from scipy.stats import chi2_contingency

table = pd.crosstab(
    payments["payment_type"],
    payments["payment_installments"]
)

chi2, p_value, dof, expected = chi2_contingency(table)

print("Chi-square:", chi2)
print("P-value:", p_value)

if p_value < 0.05:
    print("Reject H0: Payment type and number of installments are associated.")
else:
    print("Fail to reject H0: No significant association.")
```
